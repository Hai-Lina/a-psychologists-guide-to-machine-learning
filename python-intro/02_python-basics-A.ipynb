## Data types

Programming is nothing more than processing (reading, converting, outputting) data.

In a program, data is considered as individual *values*. For example, when a program asks you to enter a number (or a name, an email address, or anything else), your program reads the keyboard input and stores the entered data as a *value* in the main memory (RAM). With the appropriate command, you can then, for example, display this value on the screen.

Run the next code cell. Don’t worry: you don’t need to understand exactly what is happening yet; the example is just meant to demonstrate input (How old are you?) &rarr; processing (multiply the years by 365.25) &rarr; output (print) of data.

```python
age_years = int(input("How old are you (years)? "))   # Input
age_days = age_years * 365.25                         # Calculation
print(f"You are approximately {age_days} days old.")  # Output
```

### Why are data types important?

The data type determines how a value is represented internally.

From a programmer’s perspective, the most important thing is what you **can** and **cannot** do with a particular data type.
For example, the `upper()` method of the string type converts all characters to uppercase.

```python
"abc".upper()
```

Using `upper()` with integers, on the other hand, doesn’t make sense and will result in an error:

```python
123.upper()
```

Operators, such as mathematical operations, behave differently for different data types:

```python
1 + 1
```

```python
'abc' + 'def'
```

```python
1 + '1' # results in a "Type Error"
```

## Variables

Variables are names for values. A variable is thus a name (which we define ourselves) that points to a specific value. This allows us to refer to that value in different parts of the program using its name. This is useful because we can create a value in one place and work with it in other places.

### Assigning values

```python
firstname = 'Jana'
```

Later in the program, we can use this variable to access the value. In the following example, to display it:

```python
print(firstname)
```

### Variables Are Variable
As the name suggests, a variable can be assigned a new value any number of times during the program’s execution.

```python
firstname = 'Wolfgang Amadeus'
print(firstname)
firstname = 'Ludwig'
print(firstname)
```

Here, we first assigned the value `Wolfgang Amadeus` to the variable `firstname` and later in the program the new value `Ludwig`. From that point on, the variable `firstname` refers to the new value.

### Variables and Non-Values

Python has a special data type `NoneType`, which can take exactly one value: `None`.

```python
type(None)
```

`None` is a non-value used wherever no value is present. For example, a variable can be initialized with the value `None` and only later in the program be assigned a different value.
If you want to check whether a variable’s value is `None`, you should use the `is` operator:

```python
firstname = None   # Assignemnt
firstname is None  # Comparison
```

## Operators

### Mathematical operators

Operators apply operations to values. That is, they do something with one or more values. In the first part of today’s session, we already covered the mathematical operators `+`, `-`, `*`, and `/`.

In addition, there are the following mathematical operators:

* `**`: Exponentiation (e.g., `2 ** 8` represents 2 to the power of 8)
* `%`: Modulo (remainder of a division; e.g., `5 % 2` is 1)

### Comparison Operators

Comparison operators are used to compare two values.

For example, you can check whether two values are equal. The corresponding operator is `==` (two equal signs!). Every comparison returns a truth value (i.e., `True` or `False`).

~~~
123 == 123
True

'Graz' == 'Vienna'
False
~~~

Of course, you can also compare calculated values with each other:

~~~
100 / 4 == 75 / 3
~~~

Python has a number of comparison operators:

* `==` returns `True` if the two values are equal
* `!=` returns `True` if the two values are not equal
* `<` returns `True` if the left value is less than the right value
* `<=` returns `True` if the left value is less than or equal to the right value
* `>` returns `True` if the left value is greater than the right value
* `>=` returns `True` if the left value is greater than or equal to the right value

## Loops

When programming, it is often necessary to repeat one or more statements, for example, for each element of a sequence (e.g., for each character in a string). For this purpose, the `for` loop is ideal.

### The for Loop

The `for` loop iterates element by element through a data type consisting of multiple elements and applies the code in the *body of the loop* to each element. The loop automatically ends when there are no more elements left.

In the following example, we use a `for` loop to print each character of a string in order. After the last character, the loop ends.

```python
sentence = "I am a horse."

for i in sentence:
    print(i)
```

This construct (`for element in sequence`) works for all data types that can provide one element after another. These are called **iterables**. Python has many such iterables. Using this, you can not only iterate through the characters of a string, but also through the elements of a list, the lines of a file, or simply through a sequence of numbers:

```python
# The range function generates an object that provides numbers in order
# from a specific range: here from 1 (inclusive) to 11 (exclusive)
# We will learn exactly what a function is tomorrow

for i in range(1, 11):
    print(i)
```

### Indentation

Here we have seen for the first time a peculiarity of Python that can sometimes be confusing: logically related blocks are indicated by indentation.

~~~
for i in range(1, 11):
    print(i)
~~~

The colon at the end of the first line and the indentation of `print(i)` tell the interpreter that this is the loop body, i.e., the statement(s) to be executed in each iteration of the loop.

An indentation of 4 spaces or one tab is recommended, but in principle, any number of spaces will work.

### Loop Counter

Sometimes, in addition to the values, you also need a running counter, i.e., the number of the current iteration. A simple way to achieve this is by defining a variable `i` that is used as a counter and is incremented by 1 in each iteration of the loop.

Here is an example of a loop with a counter:

```python
i = 0

for j in range(50, 60):
    print(i, j)

    i = i + 1
```

The same can be achieved more elegantly using the built-in `enumerate()` function:

```python
for i, j in enumerate(range(50, 60)):
    print(i, j)
```

`enumerate(iterable)` provides two values for each iteration of the loop:

* the counter (i.e., the current iteration number, starting at 0)
* the actual value from the iterable

### Nested Loops
You can nest two (or more) loops inside each other (though more than two is usually not recommended, as it can produce a huge number of iterations). This allows, for example, combining all elements from two sequences. For loop counters, the variable names `i`, `j`, and `k` are commonly used (and should be used only as counters!).

```python
for i in range(1, 11):
    for ii in range(1, 11):
        # print(f'{j} x {i} = {i*j}')
        print(i * ii)
```

**Note**: Here we have used a so-called formatted string, or "f-string." This is a string (i.e., a sequence of characters) in which we can dynamically insert the values of various variables at positions indicated by `{}`. You can learn more about f-strings on [this page](https://www.geeksforgeeks.org/python/formatted-string-literals-f-strings-python/) in the subsection "f-Strings."

## Lists

A list is an ordered sequence of elements.
You can think of a list as a container that holds other objects in a specific order. A list (`list`) is thus, after strings (`str`), another *sequence type* that we will get to know.

The data type of a list element does not matter; in other words, a list can store elements of any type:      

```python
# A list of four strings
students = ['Otto', 'Anna', 'Maria', 'Franz']
students
```

```python
# A list of four floats
temperatures = [25.4, 28.1, 20.9, 26.0, 32.6]
temperatures
```

```python
# A list of seven integers
grades = [1, 2, 1, 4, 3, 3, 2]
grades
```

A list can even contain elements of different types (though this is usually not a particularly good idea):

```python
values = ['Otto', 7, True, 3.14]
values
```

### Determining the Number of List Elements
We can determine the number of elements in a list using the `len()` function:

```python
len(students)
```

### Accessing Individual Elements
We can access a single element of a list using its **index**:


```python
students[0]
```

This also works with negative index values!

```python
students[-1]
```

### Slicing
In addition, sublists can be extracted:

```python
students
```

```python
students[1:3]
```

### Iterating Through a List

Using a `for` loop, we can go through the individual elements of a list, just as we did earlier for the individual characters of a string:

```python
for stud in students:
    print(stud)
```

### Adding Elements
We can add new elements to a list at any time. The method `append(VALUE)` adds a new element to the end of the list:

```python
students = ['Otto', 'Anna', 'Maria', 'Franz']
print(students)
students.append('Otto')
print(students)
```

If needed, we can insert an element at any position. To do this, we use the `insert()` method of the `list` object. This method expects 2 parameters:

1. The position at which the value should be inserted into the list. This is 0-based, so position `0` means "at the very beginning," `1` means "after the first element."
2. The value to be inserted.

```python
students.insert(0, 'Berta')
students
```

### Removing Elements

Similarly, we can remove elements. The `pop()` method removes the last element of the list. `pop()` also returns the value of the removed element.

```python
last_stud = students.pop()
print(last_stud)
print(students)
```

### Replacing Elements
Using the index, the value of a list element can be changed at any time:

```python
students = ['Otto', 'Anna', 'Maria', 'Franz']
print(students)
students[1] = 'Berta'
print(students)
```

Here, we have changed the value of the second element from 'Anna' to 'Berta'.

### Performing Calculations with List Values

For lists containing numerical values (`int`, `float`), Python provides functions that can be applied to all values in a list:

* `max(list)` returns the largest value in the list
* `min(list)` returns the smallest value in the list
* `sum(list)` returns the sum of all values in the list

```python
values = [7, 2, 4, 1, 4, 8, 4]
print("Largest value: {}".format(max(values)))
print("Smallest value: {}".format(min(values)))
print("Sum: {}".format(sum(values)))
```
