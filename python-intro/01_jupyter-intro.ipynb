# Jupyter Notebooks

## What is Jupyter?

Jupyter is both

* an **editor** where we can write and structure text, and
* an **interpreter** where we can write and execute code.

We will use Jupyter for our programming exercises in this course.

## Create, edit and execute cells

The notebook consists of individual **cells**, which can contain either text or code.

Jupyter notebooks have two different modes:

### Edit mode

Activate edit mode by<ul>
<li>pressing <b>enter</b> while the mouse pointer is on the cell, or</li>  
<li><b>double-clicking</b> inside a cell with the mouse</li>
</ul>

### Command mode

Activate command mode by<ul>  
<li>pressing <b>esc</b> (Windows, Linux) or <b>ctrl/cmd + enter</b> while the mouse pointer is on the cell, or</li>  
<li><b>clicking play</b> at the top of the Jupyter Notebook interface.</li>  
</ul>

When the cell is in edit mode, you can modify its content just like in a regular text editing program, such as Microsoft Word. When the cell is in command mode, you can edit the notebook as a whole—for example, you can delete, move, copy, and paste entire cells—but you cannot modify the content of individual cells.

## Navigation within the Jupyter Notebook

**Navigation with the mouse:** You can select individual cells with the mouse and navigate within the notebook using the mouse wheel or the scrollbar on the right.

**Navigation with the keyboard:** Both in edit and command mode, you can navigate within the notebook using the arrow keys on the keyboard.
